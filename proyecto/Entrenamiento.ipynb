```python
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

MODELS_DIR = Path("./modelos")
MODELS_DIR.mkdir(parents=True, exist_ok=True)
MODEL_PATH = MODELS_DIR / "modelo_activo.pkl"
print("Carpeta lista para recibir el modelo de entrenamiento.")
```

```python
import joblib
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

X, y = load_iris(return_X_y=True)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

joblib.dump(model, MODEL_PATH)
print(" Modelo activo: RandomForestClassifier")
```

```python
import joblib
from sklearn.datasets import load_iris
from sklearn.svm import SVC

X, y = load_iris(return_X_y=True)

model = SVC(kernel='rbf', random_state=42)
model.fit(X, y)

# Sobreescribe el mismo archivo .pkl
joblib.dump(model, MODEL_PATH)
print(" Modelo activo: SVC")
```

```python
import joblib
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier

X, y = load_iris(return_X_y=True)

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X, y)

# Sobreescribe el mismo archivo .pkl
joblib.dump(model, MODEL_PATH)
print(" Modelo activo: KNeighborsClassifier")
```
